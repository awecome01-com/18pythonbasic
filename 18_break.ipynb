```python

#break로 반복문 끝내기

i=0
while True:
    print(i)
    i+=1
    if i == 10:
        break
```

```python
#countinue로 코드 실행 건너뛰기
for i in range(100):
    if i%2 ==0:
        continue
    print(i)
```

```python
#while 반복문에서 continue로 코드 실행 건너뛰기
i = 0
while i <100:
    i += 1
    if i % 2 ==0:
        continue
    print(i)
```

```python
i = 0
while True:
    if i % 10 != 3:
        i += 1
        continue
    if i > 73:
        break
    print(i, end='')
    i += 1
```
