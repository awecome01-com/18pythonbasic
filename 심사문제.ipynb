```python
표준 입력으로 꼭지점 개수(정수)와 선의 길이(정수)가 입력됩니다 다음 소스 코드를 완성하여 꼭지점 개수와 21단원 표준입력5 150
선의 길이에 맞는 별이그려지게 만드세요.
별을 그릴 때는 현재 위치부터 오른쪽으로 이동해서 시작해야 하며 시계 방향으로 그려야 합니다.
import turtle as t <주어짐

n, line = map(int, input().split()) <주어짐
t.shape('turtle') <주어짐
t.speed('fastest') <주어짐
for i in range(n):
    t.forward(line)
    t.right(360 / n * 2)
    t.forward(line)
    t.left(360 / n)
```

```python
표준 입력으로 정수 두 개가 입력됩니다 22단원 표준입력:1 10
첫 번째 정수부터 두 번째 정수까지를 지수로 하는 2의 거듭제곱 리스트를 출력하는 프로그램을 만드세요
단, 리스트의 두 번째 요소와 뒤에서 두 번째 요소는 삭제한 뒤 출력하세요. 출력 결과는 리스트 형태라야 합니다.
start, stop = map(int, input().split())
a = [2 ** i for i in range(start, stop + 1)] <i를 1부터 10까지 바꾸면서 2의 i제곱을 계산해 리스트로 만들어요.
del a[1] <두 번째 값인 4를 삭제해요
del a[-2] <뒤에서 두 번째 값인 512를 삭제해요
print(a)
```

```python
표준 입력으로 2차원 리스트의 가로(col)와 세로(row)가 입력되고 그 다음 줄부터 리스트의 요소로 들어갈 문자가 입력됩니다. 23단원
이때 2차원 리스트 안에서 *는 지뢰이고 .은 지뢰가 아닙니다. 지뢰가 아닌 요소에는 인접한 지뢰의 개수를 출력하는 프로그램을 만드세요
표준입력 3 3  이거버리세요
       .**
       *..
       .*.
col, row = map(int, input().split())

matrix = [] <주어짐
for i in range(row): <주어짐
    matrix.append(list(input())) <주어짐

for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*":
            print("*", end="")
        else:
            count = 0
            for y in range(i - 1, i + 2):
                for x in range(j - 1, j + 2):
                    if 0 <= y < row and 0 <= x < col:
                        if matrix[y][x] == "*":
                            count += 1
            print(count, end="")
    print()
```

```python
표준 입력으로 물품 가격 여러 개가 문자열 한 줄로 입력되고, 각 가격은 ;(세미콜론)으로 구분되어 있습니다. 24단원
입력된 가격을 높은 가격순으로 출력하는 프로그램을 만드세요 표준입력:51900;83000;158000;367500;250000;59200;128500;1304000
이때 가격은 길이를 9로 만든 뒤 오른쪽으로 정렬하고 천단위로 ,(콤마)를 넣으세요.
prices = list(map(int, input().split(";"))) <입력한 가격들을 ;로 나누고 정수로 바꿔 리스트에 저장해요.
prices.sort(reverse=True) <리스트의 가격들을 큰 숫자부터 작은 숫자 순서로 정렬해요.
for price in prices:   <정렬된 가격을 하나씩 price에 넣으며 아래 코드를 반복해요.
    print(f"{price:>9,}")  <f로 price의 값을 문자열에 넣고, 천 단위 쉼표를 붙여 전체 9칸에서 오른쪽에 맞춰 출력해요.
```

```python
표준 입력으로 문자열 여러 개와 숫자 여러 개가 두 줄로 입력되고, 첫 번째 줄은 키, 두 번째 줄은 값으로 하여 딕셔너리를 생성합니다. 
다음 코드를 완성하여 딕셔너리에서 키가 'delta'인 키-값 쌍과 값이 30인 키-값 쌍을 삭제하도록 만드세요.
25단원 표준입력:alpha bravo charlie delta echo foxtrot golf
              30 40 50 60 70 80 90
keys = input().split() <주어짐
values = map(int,input().split()) <주어짐

x=dict(zip(keys, values)) <주어짐
x = {key: value for key, value in x.items() if key != "delta" and value != 30}
print(x)
key: value → 꺼낸 이름과 값을 다시 짝짓는 부분이에요
for key, value → 이름과 값을 꺼내 담는 부분이에요.
x.items()는 이름과 값을 짝으로 꺼낼 수 있게 해요. item()딕셔너리의 키와 값을 한 쌍씩 꺼낼 수 있게 해주는 기능이에요.
```

```python
표준 입력으로 양의 정수 두 개가 입력됩니다. 다음 소스 코드를 완성하여 두 숫자의 공약수를 세트 형태로 구하도록 만드세요. 
단, 최종 결과는 공약수의 합으로 판단합니다.26단원 표준입력:100 200
x, y = map(int, input().split())
a = {i for i in range(1, x + 1) if x % i == 0} for i in range(1, 4) → 1, 2, 3을 하나씩 꺼내서
b = {i for i in range(1, y + 1) if y % i == 0} 앞의 i → 그 숫자를 그대로 담아라 {} → 담을 곳은 세트

divisor = a & b <주어짐

result = 0 <주어짐
if type(divisor) == set: <주어짐
    result = sum(divisor) <주어짐

print(result) <주어짐
```
